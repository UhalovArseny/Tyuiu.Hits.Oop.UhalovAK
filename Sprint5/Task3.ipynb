<h2 style="color:DodgerBlue">Описание проекта:</h2>

Создать базовый класс Task в C#, который будет представлять задачи внутри
проекта. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) исользуйтие в проекте коллекции, делегаты, события.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;
using System.Collections.Generic;

// Делегаты
public delegate void TaskHandler(string message);

// Базовый класс Task
public class Task
{
    public event TaskHandler TaskCompleted;
    
    public int TaskId { get; set; }
    public string TaskName { get; set; }
    public string Priority { get; set; }
    public bool IsCompleted { get; protected set; }
    
    // Новые атрибуты
    public List<string> Tags { get; set; }
    public DateTime CreatedDate { get; set; }
    
    public Task(int taskId, string taskName, string priority)
    {
        TaskId = taskId;
        TaskName = taskName;
        Priority = priority;
        Tags = new List<string>();
        CreatedDate = DateTime.Now;
    }
    
    public virtual void MarkAsComplete()
    {
        IsCompleted = true;
        TaskCompleted?.Invoke($"Задача '{TaskName}' завершена");
    }
    
    public virtual string GetTaskDetails()
    {
        return $"ID: {TaskId}, Название: {TaskName}, Приоритет: {Priority}";
    }
    
    // Новые методы
    public void AddTag(string tag) => Tags.Add(tag);
    public void DisplayTags() => Console.WriteLine($"Теги: {string.Join(", ", Tags)}");
}

// ДелегатскаяЗадача
public class DelegateTask : Task
{
    public DateTime DueDate { get; set; }
    public string AssignedBy { get; set; }
    
    // Новые атрибуты
    public Queue<string> Actions { get; set; }
    public int Progress { get; set; }
    
    public DelegateTask(int taskId, string taskName, string priority, DateTime dueDate, string assignedBy) 
        : base(taskId, taskName, priority)
    {
        DueDate = dueDate;
        AssignedBy = assignedBy;
        Actions = new Queue<string>();
    }
    
    // Новые методы
    public void AddAction(string action) => Actions.Enqueue(action);
    
    public void CompleteAction()
    {
        if (Actions.Count > 0)
        {
            var action = Actions.Dequeue();
            Console.WriteLine($"Выполнено: {action}");
        }
    }
    
    public override string GetTaskDetails()
    {
        return base.GetTaskDetails() + $", Срок: {DueDate:dd.MM}, Назначил: {AssignedBy}";
    }
}

// КоманднаяЗадача
public class TeamTask : Task
{
    public List<string> Members { get; set; }
    
    // Новые атрибуты
    public string TeamLead { get; set; }
    public Dictionary<string, string> Roles { get; set; }
    
    public TeamTask(int taskId, string taskName, string priority, string teamLead) 
        : base(taskId, taskName, priority)
    {
        TeamLead = teamLead;
        Members = new List<string>();
        Roles = new Dictionary<string, string>();
    }
    
    // Новые методы
    public void AddMember(string name, string role)
    {
        Members.Add(name);
        Roles[name] = role;
    }
    
    public void ShowTeam()
    {
        Console.WriteLine($"Команда '{TaskName}':");
        foreach (var member in Members)
            Console.WriteLine($"  - {member} ({Roles[member]})");
    }
    
    public override string GetTaskDetails()
    {
        return base.GetTaskDetails() + $", Участников: {Members.Count}";
    }
}

// Демонстрация
var task1 = new Task(1, "Базовая", "High");
var task2 = new DelegateTask(2, "Делегатская", "Medium", DateTime.Now.AddDays(3), "Менеджер");
var task3 = new TeamTask(3, "Командная", "Low", "Тимлид");

// События
task1.TaskCompleted += (msg) => Console.WriteLine(msg);
task2.TaskCompleted += (msg) => Console.WriteLine(msg);

// Коллекции
task1.AddTag("важно");
task1.AddTag("срочно");

task2.AddAction("Подготовить");
task2.AddAction("Согласовать");
task2.CompleteAction();

task3.AddMember("Анна", "Разработчик");
task3.AddMember("Петр", "Тестировщик");

// Вывод
Console.WriteLine(task1.GetTaskDetails());
Console.WriteLine(task2.GetTaskDetails());
Console.WriteLine(task3.GetTaskDetails());

task1.DisplayTags();
task3.ShowTeam();

task1.MarkAsComplete();
```
